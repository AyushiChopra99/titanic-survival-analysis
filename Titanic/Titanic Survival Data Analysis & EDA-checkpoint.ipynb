```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None)
sns.set_theme(style="whitegrid")

print(" Libraries loaded")
```

```python
df = pd.read_csv("Titanic-Dataset.csv")

print("Shape:", df.shape)
print("\nColumns:", df.columns.tolist())
df.head()
```

```python
# Fill Age nulls with median age
df["Age"] = df["Age"].fillna(df["Age"].median())

if "Cabin" in df.columns:
    df.drop(columns=["Cabin"], inplace=True)

df.dropna(subset=["Embarked"], inplace=True)

cols_to_drop = ["PassengerId", "Name", "Ticket"]
cols_to_drop = [col for col in cols_to_drop if col in df.columns]
df.drop(columns=cols_to_drop, inplace=True)

print("Cleaning complete")
print("Final shape:", df.shape)
print("Remaining nulls:", df.isnull().sum().sum())
```

```python
df["AgeGroup"] = pd.cut(df["Age"],
                         bins=[0, 12, 18, 35, 60, 100],
                         labels=["Child", "Teen", "Young Adult", "Adult", "Senior"])

print("Age Group column added")
df[["Age", "AgeGroup"]].head(10)
```

```python
print("=== Survival Summary ===")
total = len(df)
survived = df["Survived"].sum()
print(f"Total Passengers : {total}")
print(f"Survived         : {survived} ({round(survived/total*100, 1)}%)")
print(f"Did Not Survive  : {total - survived} ({round((total-survived)/total*100, 1)}%)")
print(f"Average Age      : {df['Age'].mean():.1f} years")
print(f"Average Fare     : £{df['Fare'].mean():.2f}")
```

```python
plt.figure(figsize=(8, 5))
sns.countplot(x="Survived", data=df,
              hue="Survived", palette=["tomato", "steelblue"],
              legend=False)
plt.xticks([0, 1], ["Did Not Survive", "Survived"])
plt.title("Overall Survival Count", fontsize=16, fontweight="bold")
plt.xlabel("Outcome")
plt.ylabel("Number of Passengers")
plt.tight_layout()
plt.savefig("chart1_survival_count.png", dpi=150)
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.countplot(x="Sex", data=df, hue="Survived",
              palette=["tomato", "steelblue"])
plt.title("Survival by Gender", fontsize=16, fontweight="bold")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.legend(["Did Not Survive", "Survived"])
plt.tight_layout()
plt.savefig("chart2_survival_by_gender.png", dpi=150)
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.countplot(x="Pclass", data=df, hue="Survived",
              palette=["tomato", "steelblue"])
plt.title("Survival by Passenger Class", fontsize=16, fontweight="bold")
plt.xlabel("Passenger Class (1 = First, 3 = Third)")
plt.ylabel("Count")
plt.legend(["Did Not Survive", "Survived"])
plt.tight_layout()
plt.savefig("chart3_survival_by_class.png", dpi=150)
plt.show()
```

```python
plt.figure(figsize=(12, 5))
sns.histplot(data=df, x="Age", hue="Survived",
             bins=30, kde=True,
             palette=["tomato", "steelblue"])
plt.title("Age Distribution by Survival", fontsize=16, fontweight="bold")
plt.xlabel("Age")
plt.ylabel("Count")
plt.legend(["Did Not Survive", "Survived"])
plt.tight_layout()
plt.savefig("chart4_age_distribution.png", dpi=150)
plt.show()
```

```python
df.to_csv("cleaned_titanic_data.csv", index=False)
print("Saved: cleaned_titanic_data.csv")
print("Final shape:", df.shape)
```

```python
""" Key Findings

1. Only **38.4%** of passengers survived the Titanic disaster
2. **Female passengers** had a significantly higher survival rate (~74%) vs males (~19%)
3. **First class passengers** had the highest survival rate — nearly 63%
4. **Children under 12** had better survival odds than adults
5. Passengers who paid **higher fares** (First Class) were prioritized for lifeboats

Files Saved
- `cleaned_titanic_data.csv`
- `chart1_survival_count.png`
- `chart2_survival_by_gender.png`
- `chart3_survival_by_class.png`
- `chart4_age_distribution.png`"""
```
